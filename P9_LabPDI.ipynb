# **Practica 9.- Mejorar Calidad de Imagen a Color**

En la siguiente practica se realizará una mejora en la calidad de una imagen a color mediante la ecualización del canal value de la representación HSV de la imagen. Para empezar, se importan las librerías necesarias.

```python
import cv2
import matplotlib.pyplot as plt
```

Después, cargamos nuestra primera imagen mediante en color y la convertimos a HSV para posteriormente aislar cada canal.

```python
# Cargar imagen oscura
img_bgr = cv2.imread('Im004_ACRIMA.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# Convertir a HSV
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

# Separar canales H, S y V
h, s, v = cv2.split(img_hsv)
```

Realizamos una ecualización de unicamente el canal V (brillo) y posteriormente volvemos a combinar los canales con el nuevo valor V. La imagen es convertida de nuevo a BGR y luego RGB y es visualizada para comparar.

```python
# Ecualizar solo el canal de valor
v_eq = cv2.equalizeHist(v)

# Combinar canales con el V nuevo
img_hsv_eq = cv2.merge([h, s, v_eq])

# Convertir de nuevo RGB
img_eq_bgr = cv2.cvtColor(img_hsv_eq, cv2.COLOR_HSV2BGR)
img_eq_rgb = cv2.cvtColor(img_eq_bgr, cv2.COLOR_BGR2RGB)
```

```python
# Mostrar imágenes original y mejorada
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Imagen original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_eq_rgb)
plt.title('Imagen mejorada')
plt.axis('off')

plt.tight_layout()
plt.show()
```

El mismo proceso se repetirá para otras dos imagenes médicas distintas.

```python
import cv2
import matplotlib.pyplot as plt

# Cargar imagen
img1_bgr = cv2.imread('Colonoscopy2.jpg')
img2_bgr = cv2.imread('histology.jpeg')

# Convertir a RGB para visualizar
img1_rgb = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2_bgr, cv2.COLOR_BGR2RGB)

# Convertir a HSV
img1_hsv = cv2.cvtColor(img1_bgr, cv2.COLOR_BGR2HSV)
img2_hsv = cv2.cvtColor(img2_bgr, cv2.COLOR_BGR2HSV)

# Separar canales
h1, s1, v1 = cv2.split(img1_hsv)
h2, s2, v2 = cv2.split(img2_hsv)

# Ecualizar canal V
v1_eq = cv2.equalizeHist(v1)
v2_eq = cv2.equalizeHist(v2)

# Combinar canales nuevamente
img1_hsv_eq = cv2.merge([h1, s1, v1_eq])
img2_hsv_eq = cv2.merge([h2, s2, v2_eq])

# Convertir a BGR y luego RGB
img1_eq_bgr = cv2.cvtColor(img1_hsv_eq, cv2.COLOR_HSV2BGR)
img1_eq_rgb = cv2.cvtColor(img1_eq_bgr, cv2.COLOR_BGR2RGB)
img2_eq_bgr = cv2.cvtColor(img2_hsv_eq, cv2.COLOR_HSV2BGR)
img2_eq_rgb = cv2.cvtColor(img2_eq_bgr, cv2.COLOR_BGR2RGB)

# Mostrar Resultados
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img1_rgb)
plt.title('Imagen 1 Original')
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(img1_eq_rgb)
plt.title('Imagen 1 Mejorada')
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(img2_rgb)
plt.title('Imagen 2 Original')
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(img2_eq_rgb)
plt.title('Imagen 2 Mejorada')
plt.axis('off')

plt.tight_layout()

plt.show()
```
